Exercício 1 — Contagem

```python
limite = int(input("Digite um número: "))
contador = 1

while contador <= limite:
    print(contador)
    contador += 1
```

Exercício 2 — Contagem regressiva

```python
numero = int(input("Digite um número: "))

while numero >= 0:
    print(numero)
    numero -= 1
```

Exercício 3 — Números pares

```python
numero = 0

while numero <= 20:
    print(numero)
    numero += 2
```

Exercício 4 — Soma dos números

```python
limite = int(input("Digite um número: "))
soma = 0
contador = 1

while contador <= limite:
    soma += contador
    contador += 1

print(f"Soma: {soma}")
```

Exercício 5 — Tabuada

```python
numero = int(input("Digite um número: "))
contador = 1

while contador <= 10:
    resultado = numero * contador
    print(f"{numero} x {contador} = {resultado}")
    contador += 1
```

Exercício 6 — Senha

```python
senha = ""

while senha != "1234":
    senha = input("Digite a senha: ")
    if senha != "1234":
        print("Senha incorreta!")

print("Acesso permitido!")
```

Exercício 7 — Somando até digitar zero

```python
soma = 0
numero = int(input("Digite um número: "))

while numero != 0:
    soma += numero
    numero = int(input("Digite um número: "))

print(f"Soma: {soma}")
```

Exercício 8 — Maior número

```python
maior = 0
numero = int(input("Digite um número: "))

while numero != 0:
    if numero > maior:
        maior = numero
    numero = int(input("Digite um número: "))

print(f"Maior número: {maior}")
```

Exercício 9 — Média das notas

```python
contador = 1
soma_notas = 0.0

while contador <= 5:
    nota = float(input("Digite a nota: "))
    soma_notas += nota
    contador += 1

media = soma_notas / 5
print(f"Média: {media:.1f}")
```

Exercício 10 — Caixa eletrônico

```python
saldo = 1000.0
opcao = 0

while opcao != 4:
    print("\n--- MENU ---")
    print("1 - Depositar")
    print("2 - Sacar")
    print("3 - Ver saldo")
    print("4 - Sair")
    opcao = int(input("Escolha uma opção: "))

    if opcao == 1:
        valor = float(input("Digite o valor para depósito: R$ "))
        saldo += valor
        print(f"Depósito realizado com sucesso!")
    elif opcao == 2:
        valor = float(input("Digite o valor para saque: R$ "))
        if valor <= saldo:
            saldo -= valor
            print(f"Saque realizado com sucesso!")
        else:
            print("Saldo insuficiente!")
    elif opcao == 3:
        print(f"Saldo atual: R$ {saldo:.2f}")
    elif opcao == 4:
        print("Saindo do sistema. Até logo!")
    else:
        print("Opção inválida! Tente novamente.")
```
